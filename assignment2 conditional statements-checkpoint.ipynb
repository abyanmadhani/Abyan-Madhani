```python
balance=50000
while True :
    print("\n======= ATM MENU =======")
    print("1. check balance")
    print("2. deposite")
    print("3. withdraw")
    print("4. exit")

    choice=int(input("enter your choice(1-4):"))
    if choice==1 :
        print(f"your balance is rs {balance}")
    elif choice==2 :
        amount=int(input("enter your deposite amount"))
        balance+=amount
        print(f"rs {amount} deposited successfully,current balance is rs {balance}")
    elif choice==3 :
        cash=int(input("enter the amount of cash you want to withdraw"))
        if cash > balance:
            print("Insufficient balance! Withdrawal denied.")
        else:
             balance-=cash
             print(f"rs {cash} has been withdrawl from your account. current balance remaining is {balance}")
    elif choice==4 :
        print("thanks for using")
        break

    else:
        print("invalid choice.\n TRY AGAIN")
        
        
        
    
        
```

```python
passed=0
failed=0
count=0
while True :
    number=input("ENTER STUDENT NAME (to stop the process you must enter -1:)")
    if number=="-1" :
        print("list is completed")
        break
    else:
        print("list not over yet")

    marks=int(input(f"enter marks for student {number}:"))
    count+=1 

    if marks>=50 :
        print(f"{number}: passed:")
        passed+=1
    else:
        print(f"{number}: failed:")
        failed+=1

print(f"\ncount Students: {count}")
print(f"count Passed: {passed}")
print(f"count Failed: {failed}")
```

```python

total = 0
items = 0

burger_quantity = 0
pizza_quantity = 0
fries_quantity = 0
drink_quantity = 0

while True:
    print("\n===== RESTAURANT MENU =====")
    print("1. Burger - Rs. 500")
    print("2. Pizza  - Rs. 1000")
    print("3. Fries  - Rs. 300")
    print("4. Drink  - Rs. 150")
    print("5. Checkout")

    choice = int(input("Enter your choice (1-5): "))

    if choice == 1:
        quantity = int(input("How many Burgers? "))
        burger_quantity += quantity
        print(f"{quantity} Burger(s) added")
        total += 500 * quantity
        items += quantity

    elif choice == 2:
        quantity = int(input("How many Pizzas? "))
        pizza_quantity += quantity
        print(f"{quantity} Pizza(s) added")
        total += 1000 * quantity
        items += quantity

    elif choice == 3:
        quantity = int(input("How many Fries? "))
        fries_quantity += quantity
        print(f"{quantity} Fries added")
        total += 300 * quantity
        items += quantity

    elif choice == 4:
        quantity = int(input("How many Drinks? "))
        drink_quantity += quantity
        print(f"{quantity} Drink(s) added")
        total += 150 * quantity
        items += quantity

    elif choice == 5:
        print("\n======= FINAL RECEIPT =======")
        if burger_quantity > 0:
            print(f"Burger x {burger_quantity} = Rs. {500 * burger_quantity}")
        if pizza_quantity > 0:
            print(f"Pizza  x {pizza_quantity} = Rs. {1000 * pizza_quantity}")
        if fries_quantity > 0:
            print(f"Fries  x {fries_quantity} = Rs. {300 * fries_quantity}")
        if drink_quantity > 0:
            print(f"Drink  x {drink_quantity} = Rs. {150 * drink_quantity}")
        print(f"\nTotal Items: {items}")
        print(f"Total Bill = Rs. {total}")
        print("Thank you for ordering!")
        break

    else:
        print("Invalid choice. Please try again.")
```
